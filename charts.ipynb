```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
funnel = pd.DataFrame({
    'step': ['Viewed product', 'Added to cart', 'Purchased'],
    'sessions': [681995, 431205, 181995]
})
ax = funnel.plot(kind='barh', x='step', y='sessions', legend=False,
                 title='Purchase funnel: 58% of carts are abandoned')
ax.invert_yaxis()
ax.bar_label(ax.containers[0], fmt='{:,.0f}')
ax.set_xlim(0, 800000)
ax.set_ylabel('')
plt.savefig('chart1_funnel.png', dpi=200, bbox_inches='tight')
```

```python
monthly = pd.read_csv('02_monthly_revenue.csv', parse_dates=['month'])
monthly = monthly.iloc[:-1]
ax = monthly.plot(x='month', y='revenue', legend=False,
                  title='Monthly revenue (excluding cancelled and returned items)')
ax.set_ylabel('Revenue ($)')
plt.savefig('chart2_monthly_revenue.png', dpi=200, bbox_inches='tight')
```

```python
cats = pd.read_csv('05_category_revenue_returns.csv')
cats['revenue_per_order'] = cats['revenue'] / cats['orders']
top10 = cats.head(10).sort_values('revenue_per_order')
ax = top10.plot(kind='barh', x='category', y='revenue_per_order', legend=False,
                title='Revenue per order, top 10 categories by total revenue')
ax.set_xlabel('Revenue per order ($)')
ax.set_ylabel('')
plt.savefig('chart3_category_value.png', dpi=200, bbox_inches='tight')
```
